```python
# ============================================================
# 1. DATA LOADING AND EXPLORATION
# ============================================================

import pandas as pd
import numpy as np

df = pd.read_csv("train.csv")

print(df.head())

print("\nDimensions")
print(df.shape)

print("\nTypes de données")
print(df.dtypes)

print("\nValeurs manquantes")
print(df.isnull().sum())
```

```python
print(df.describe())
```

```python
# ============================================================
# 2. DATA CLEANING
# ============================================================

print(df.isnull().sum())

# Suppression des doublons
df.drop_duplicates(inplace=True)

print("Dimensions après nettoyage")
print(df.shape)
```

```python
# ============================================================
# 3. STATISTICAL ANALYSIS
# ============================================================

from scipy import stats
```

```python
for col in df.columns:

    print("\n", col)

    print("Mean :", np.mean(df[col]))
    print("Median :", np.median(df[col]))
    print("Mode :", stats.mode(df[col], keepdims=True))
```

```python
for col in df.columns:

    print("\n", col)

    print("Range :", np.max(df[col]) - np.min(df[col]))

    print("Variance :", np.var(df[col]))

    print("Std :", np.std(df[col]))
```

```python
skewness = df.apply(stats.skew)

print(skewness.sort_values())
```

```python
kurt = df.apply(stats.kurtosis)

print(kurt.sort_values())
```

```python
corr_price = df.corr()["price_range"]

print(
    corr_price.sort_values(
        ascending=False
    )
)
```

```python
from scipy.stats import pearsonr

for col in df.columns:

    if col != "price_range":

        corr, p = pearsonr(
            df[col],
            df["price_range"]
        )

        print(
            col,
            "corr =",
            round(corr,3),
            "p =",
            round(p,5)
        )
```

```python
group0 = df[df.price_range==0]["ram"]
group1 = df[df.price_range==1]["ram"]

t,p = stats.ttest_ind(
    group0,
    group1
)

print(t,p)
```

```python
g0 = df[df.price_range==0]["ram"]
g1 = df[df.price_range==1]["ram"]
g2 = df[df.price_range==2]["ram"]
g3 = df[df.price_range==3]["ram"]

f,p = stats.f_oneway(
    g0,g1,g2,g3
)

print("F =",f)
print("P =",p)
```

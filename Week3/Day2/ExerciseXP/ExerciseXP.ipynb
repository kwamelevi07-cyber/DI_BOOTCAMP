```python
# EXERCICE 1 : Opérations matricielles

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Création d'une matrice 3x3
matrix = np.array([
    [4, 7, 2],
    [3, 6, 1],
    [2, 5, 9]
])

print("Matrice :\n", matrix)

# Déterminant
det = np.linalg.det(matrix)
print("\nDéterminant :", det)

# Inverse de la matrice
if det != 0:
    inv_matrix = np.linalg.inv(matrix)
    print("\nInverse de la matrice :\n", inv_matrix)
else:
    print("\nLa matrice n'a pas d'inverse")
```

```python
# EXERCICE 2 : Analyse statistique

data = np.random.randn(50) * 10 + 50  # données centrées autour de 50

mean = np.mean(data)
median = np.median(data)
std = np.std(data)

print("Données :", data)
print("\nMoyenne :", mean)
print("Médiane :", median)
print("Écart type :", std)
```

```python
# EXERCICE 4 : Manipulation de données avec NumPy et Pandas

data = np.random.randint(1, 100, size=(5, 4))

df = pd.DataFrame(data, columns=["A", "B", "C", "D"])

print("DataFrame :\n", df)

# Sélection conditionnelle
filtered = df[df > 50]

print("\nValeurs > 50 :\n", filtered)

# Agrégations
print("\nSomme des colonnes :\n", df.sum())
print("\nMoyenne des colonnes :\n", df.mean())
```

```python
# EXERCICE 5 : Représentation d'images

# Une image en NumPy = matrice (n x m) pour niveaux de gris

image = np.array([
    [0, 50, 100, 150, 200],
    [10, 60, 110, 160, 210],
    [20, 70, 120, 170, 220],
    [30, 80, 130, 180, 230],
    [40, 90, 140, 190, 255]
])

print("Image en niveaux de gris (5x5) :\n", image)

plt.imshow(image, cmap='gray')
plt.title("Image 5x5 en niveaux de gris")
plt.colorbar()
plt.show()
```

```python
# EXERCICE 6 : Tests d'hypothèses de base

np.random.seed(42)

productivity_before = np.random.normal(loc=50, scale=10, size=30)
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

mean_before = np.mean(productivity_before)
mean_after = np.mean(productivity_after)

print("Moyenne avant formation :", mean_before)
print("Moyenne après formation :", mean_after)

if mean_after > mean_before:
    print("Hypothèse : Le programme améliore la productivité ✔️")
else:
    print("Hypothèse : Pas d'amélioration ")
```

```python
# EXERCICE 7 : Comparaison de tableaux complexes

a = np.array([10, 25, 30, 40, 50])
b = np.array([15, 20, 35, 30, 60])

result = a > b

print("Tableau A :", a)
print("Tableau B :", b)
print("A > B :", result)
```

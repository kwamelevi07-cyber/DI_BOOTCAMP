Daily Challenge: NumPy, Pandas, and Matplotlib Integration

```python
# importation des bibliothèques nécessaires
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

```

```python
# 1.Data Preparation
# Pour obtenir les mêmes résultats à chaque exécution
np.random.seed(42)

# Liste des villes
cities = [
    "Abidjan",
    "Paris",
    "New York",
    "Tokyo",
    "London",
    "Moscow",
    "Sydney",
    "Cairo",
    "Rio",
    "Toronto"
]

# Liste des mois
months = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

# Génération des températures aléatoires entre -5°C et 35°C
temperature_data = np.random.uniform(
    low=-5,
    high=35,
    size=(10, 12)
)

# Création du DataFrame
df = pd.DataFrame(
    temperature_data,
    index=cities,
    columns=months
)

# Affichage des données
print("Dataset des températures :")
display(df.round(2))
```

```python
# 2.Data Analysis
annual_average = df.mean(axis=1)

print("Température moyenne annuelle par ville :")
display(annual_average.round(2))

highest_city = annual_average.idxmax()
lowest_city = annual_average.idxmin()

highest_temp = annual_average.max()
lowest_temp = annual_average.min()

print(f"Ville la plus chaude : {highest_city} ({highest_temp:.2f} °C)")
print(f"Ville la plus froide : {lowest_city} ({lowest_temp:.2f} °C)")
```

```python
# 3.Data Visualization
plt.figure(figsize=(14, 8))

for city in df.index:
    plt.plot(
        months,
        df.loc[city],
        marker='o',
        linewidth=2,
        label=city
    )

plt.title("Average Monthly Temperature Trends by City")
plt.xlabel("Months")
plt.ylabel("Temperature (°C)")
plt.grid(True)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()

plt.show()
```

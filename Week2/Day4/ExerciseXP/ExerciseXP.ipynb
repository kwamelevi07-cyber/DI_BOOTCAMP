```python
# Exercice 2 : Variation de température sur une semaine

import matplotlib.pyplot as plt

jours = [
    "Lundi",
    "Mardi",
    "Mercredi",
    "Jeudi",
    "Vendredi",
    "Samedi",
    "Dimanche"
]

temperatures = [72, 74, 76, 80, 82, 78, 75]

plt.figure(figsize=(8,5))

plt.plot(
    jours,
    temperatures,
    marker='o',
    linewidth=2
)

plt.xlabel("Jour")
plt.ylabel("Température (°F)")
plt.title("Variation de la température sur une semaine")

plt.grid(True)

plt.show()

```

```python
# Exercice 3 : Visualisation des ventes mensuelles

import matplotlib.pyplot as plt

mois = [
    "Janvier",
    "Février",
    "Mars",
    "Avril",
    "Mai"
]

ventes = [
    5000,
    5500,
    6200,
    7000,
    7500
]

plt.figure(figsize=(8,5))

plt.bar(
    mois,
    ventes
)

plt.xlabel("Mois")
plt.ylabel("Montant des ventes ($)")
plt.title("Ventes mensuelles du magasin")

plt.show()
```

```python
# Exercice 4 : Distribution des moyennes cumulatives (CGPA)

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Student Mental health.csv")

plt.figure(figsize=(10,6))

sns.histplot(
    data=df,
    x="What is your CGPA?",
    color="skyblue"
)

plt.title("Distribution des catégories de CGPA")

plt.xlabel("CGPA")
plt.ylabel("Nombre d'étudiants")

plt.show()
```

```python
# Exercice 5 : Comparaison des niveaux d'anxiété selon le sexe

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Student Mental health.csv")

plt.figure(figsize=(10,6))

sns.countplot(
    data=df,
    x="Choose your gender",
    hue="Do you have Anxiety?",
    palette="Set2"
)

plt.title("Comparaison des niveaux d'anxiété selon le sexe")

plt.xlabel("Sexe")
plt.ylabel("Nombre d'étudiants")

plt.show()
```

```python
# Exercice 6 : Relation entre l'âge et les crises de panique

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Student Mental health.csv")

# Conversion Oui/Non vers 1/0

df["Panic_Attack_Numeric"] = (
    df["Do you have Panic attack?"]
    .map({
        "Yes": 1,
        "No": 0
    })
)

plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="Age",
    y="Panic_Attack_Numeric",
    alpha=0.7
)

plt.title("Relation entre l'âge et les crises de panique")

plt.xlabel("Âge")
plt.ylabel("Crise de panique")

plt.yticks(
    [0,1],
    ["Non","Oui"]
)

plt.show()
```

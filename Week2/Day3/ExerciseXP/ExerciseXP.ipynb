```python
# EXERCICE 2 : Statistiques descriptives


import numpy as np
from scipy import stats

data = [12, 15, 13, 12, 18, 20, 22, 21]

mean = np.mean(data)
median = np.median(data)
variance = np.var(data)
std_dev = np.std(data)

print("Moyenne :", mean)
print("Médiane :", median)
print("Variance :", variance)
print("Écart type :", std_dev)
```

```python
# EXERCICE 3 : Comprendre les distributions


import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

x = np.linspace(10, 90, 1000)

y = norm.pdf(x, loc=50, scale=10)

plt.figure(figsize=(8,5))
plt.plot(x, y)

plt.title("Distribution Normale (μ=50, σ=10)")
plt.xlabel("Valeurs")
plt.ylabel("Densité")

plt.grid(True)
plt.show()
```

```python
# EXERCICE 4 : Application du test t


import numpy as np
from scipy.stats import ttest_ind

np.random.seed(42)

data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

t_stat, p_value = ttest_ind(data1, data2)

print("Statistique t :", t_stat)
print("Valeur p :", p_value)

if p_value < 0.05:
    print("Différence significative entre les deux groupes")
else:
    print("Aucune différence significative")
```

```python
# EXERCICE 5 : Analyse de régression linéaire


from scipy.stats import linregress

house_sizes = [50, 70, 80, 100, 120]
house_prices = [150000, 200000, 210000, 250000, 280000]

slope, intercept, r_value, p_value, std_err = linregress(
    house_sizes,
    house_prices
)

print("Pente :", slope)
print("Ordonnée à l'origine :", intercept)

# Prix estimé pour 90 m²
predicted_price = slope * 90 + intercept

print("Prix estimé pour 90 m² :", predicted_price)

print("\nInterprétation :")
print(
    f"Chaque mètre carré supplémentaire augmente le prix d'environ {slope:.2f} unités monétaires."
)
```

```python
# EXERCICE 6 : Comprendre l'ANOVA


from scipy.stats import f_oneway

fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

f_stat, p_value = f_oneway(
    fertilizer_1,
    fertilizer_2,
    fertilizer_3
)

print("Valeur F :", f_stat)
print("Valeur p :", p_value)

if p_value < 0.05:
    print("Les engrais ont des effets significativement différents.")
else:
    print("Aucune différence significative entre les engrais.")

print("\nExplication :")
print(
    "Si p > 0.05, on ne peut pas rejeter l'hypothèse nulle. "
    "On considère alors que les différences observées peuvent être dues au hasard."
)
```

```python
# EXERCICE 7 : Distribution binomiale


from scipy.stats import binom

# Probabilité d'obtenir exactement 5 faces sur 10 lancers

n = 10
p = 0.5
k = 5

probability = binom.pmf(k, n, p)

print("Probabilité d'obtenir exactement 5 faces :", probability)
```

```python
# EXERCICE 8 : Coefficients de corrélation


import pandas as pd
from scipy.stats import pearsonr, spearmanr

data = pd.DataFrame({
    'age': [23, 25, 30, 35, 40],
    'income': [35000, 40000, 50000, 60000, 70000]
})

pearson_corr, pearson_p = pearsonr(
    data["age"],
    data["income"]
)

spearman_corr, spearman_p = spearmanr(
    data["age"],
    data["income"]
)

print("Corrélation de Pearson :", pearson_corr)
print("p-value Pearson :", pearson_p)

print("\nCorrélation de Spearman :", spearman_corr)
print("p-value Spearman :", spearman_p)
```

```python
import pandas as pd
import numpy as np

# Chargement du dataset
df = pd.read_csv(
    "Airplane_Crashes_and_Fatalities_Since_1908_t0_2023.csv",
    encoding="latin1"
)

# Aperçu
print(df.head())
print(df.info())

# Conversion des dates
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# Variables numériques
numeric_cols = [
    "Aboard",
    "Fatalities",
    "Ground"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Gestion des valeurs manquantes
df["Fatalities"] = df["Fatalities"].fillna(0)
df["Aboard"] = df["Aboard"].fillna(df["Aboard"].median())

# Création de nouvelles variables
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Decade"] = (df["Year"] // 10) * 10

# Taux de survie
df["SurvivalRate"] = (
    (df["Aboard"] - df["Fatalities"]) /
    df["Aboard"]
)

print(df.isnull().sum())
```

2. Analyse exploratoire des données

```python
total_accidents = len(df)
print("Nombre total d'accidents :", total_accidents)
```

```python
total_deaths = df["Fatalities"].sum()
print("Nombre total de décès :", total_deaths)
```

```python
avg_survival = df["SurvivalRate"].mean()

print("Taux moyen de survie :", avg_survival)
```

```python
accidents_per_year = df.groupby("Year").size()

print(accidents_per_year.head())
```

```python
deaths_per_decade = (
    df.groupby("Decade")["Fatalities"]
      .sum()
      .sort_index()
)

print(deaths_per_decade)
```

3. Analyse statistique

```python
from scipy import stats

mean_deaths = df["Fatalities"].mean()
print("Moyenne :", mean_deaths)

median_deaths = df["Fatalities"].median()
print("Médiane :", median_deaths)

std_deaths = df["Fatalities"].std()
print("Écart-type :", std_deaths)

print(df["Fatalities"].describe())
```

Test d'hypothèse

```python
deaths_1910 = df[df["Decade"] == 1910]["Fatalities"]
deaths_2020 = df[df["Decade"] == 2020]["Fatalities"]

t_stat, p_value = stats.ttest_ind(
    deaths_1910,
    deaths_2020,
    equal_var=False,
    nan_policy="omit"
)

print("t =", t_stat)
print("p =", p_value)
```

Interprétation

```python
alpha = 0.05

if p_value < alpha:
    print("Différence significative")
else:
    print("Pas de différence significative")
```

```python
# Exercice 3 : Régression Logistique avec GridSearchCV

from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression

param_grid = {
    'C':[0.01,0.1,1,10,100],
    'penalty':['l1','l2'],
    'solver':['liblinear']
}

grid_lr = GridSearchCV(
    LogisticRegression(max_iter=1000),
    param_grid,
    cv=5,
    scoring='accuracy'
)

grid_lr.fit(X_train_scaled, y_train)
#Meilleurs paramètres
print(grid_lr.best_params_)
print(grid_lr.best_score_)
#Évaluation
best_lr = grid_lr.best_estimator_

y_pred = best_lr.predict(X_test_scaled)

print("Accuracy :", accuracy_score(y_test,y_pred))
```

```python
# Exercice 4 : SVM sans Grid Search
#Entraînement
from sklearn.svm import SVC

svm_model = SVC(
    kernel='rbf',
    C=1,
    gamma='scale'
)

svm_model.fit(X_train_scaled, y_train)
# Évaluation
y_pred_svm = svm_model.predict(X_test_scaled)

print("Accuracy :", accuracy_score(y_test,y_pred_svm))
print(classification_report(y_test,y_pred_svm))

"""Justification
RBF est le noyau le plus utilisé.
C=1 fournit un bon compromis biais/variance.
gamma='scale' est recommandé par Scikit-Learn."""
```

```python
# Exercice 5 : SVM avec Grid Search

param_grid_svm = {
    'C':[0.1,1,10,100],
    'gamma':[1,0.1,0.01,0.001],
    'kernel':['rbf','linear']
}

grid_svm = GridSearchCV(
    SVC(),
    param_grid_svm,
    cv=5,
    scoring='accuracy'
)

grid_svm.fit(X_train_scaled, y_train)

# Résultats

print(grid_svm.best_params_)
print(grid_svm.best_score_)

# Test

best_svm = grid_svm.best_estimator_

y_pred = best_svm.predict(X_test_scaled)

print("Accuracy :", accuracy_score(y_test,y_pred))
```

```python
# Exercice 7 : XGBoost avec Grid Search

param_grid_xgb = {
    'n_estimators':[50,100,200],
    'max_depth':[3,5,7],
    'learning_rate':[0.01,0.1,0.2]
}

grid_xgb = GridSearchCV(
    XGBClassifier(
        objective='binary:logistic',
        random_state=42
    ),
    param_grid_xgb,
    cv=5,
    scoring='accuracy'
)

grid_xgb.fit(X_train, y_train)

# Meilleurs paramètres

print(grid_xgb.best_params_)
print(grid_xgb.best_score_)

# Évaluation finale

best_xgb = grid_xgb.best_estimator_

y_pred = best_xgb.predict(X_test)

print("Accuracy :", accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))

"""Conclusion à mettre dans le rapport
La régression logistique fournit une base de comparaison simple et interprétable.
Le SVM améliore souvent les performances grâce à sa capacité à gérer des frontières de décision complexes.
XGBoost est généralement le modèle le plus performant grâce à son mécanisme de boosting.
L'utilisation de GridSearchCV permet d'optimiser automatiquement les hyperparamètres et d'obtenir de meilleures performances que les paramètres choisis manuellement."""
```

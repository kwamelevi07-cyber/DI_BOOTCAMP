Exercice 1 : Définition du problème et collecte de données pour la prédiction des défauts de paiement des prêts
1. Énoncé du problème

Les institutions financières accordent chaque année un grand nombre de prêts à des particuliers et à des entreprises. Cependant, certains emprunteurs ne remboursent pas leurs prêts dans les délais prévus, ce qui entraîne des pertes financières importantes.

L'objectif de ce projet est de développer un modèle d'apprentissage automatique capable de prédire si un demandeur de prêt présente un risque élevé de défaut de paiement à partir de ses caractéristiques financières, personnelles et historiques.

Cette prédiction permettra :

de réduire les pertes financières ;
d'améliorer la gestion des risques ;
d'accélérer le processus d'approbation des prêts ;
de prendre des décisions plus objectives et cohérentes.

Le problème est donc un problème de classification binaire :

Classe 1 : le client rembourse correctement son prêt.
Classe 0 : le client risque de faire défaut.

Variables les plus pertinentes
1. Credit_History

C'est généralement la variable la plus importante.

Pourquoi ?

Un client ayant déjà remboursé correctement ses crédits est moins risqué.
Les banques utilisent fortement cet indicateur.
2. ApplicantIncome

Pourquoi ?

Plus le revenu est élevé, plus la capacité de remboursement est importante.
3. CoapplicantIncome

Pourquoi ?

Un revenu supplémentaire réduit le risque financier.
4. LoanAmount

Pourquoi ?

Les prêts élevés sont souvent plus difficiles à rembourser.
5. Loan_Amount_Term

Pourquoi ?

Une durée plus longue réduit les mensualités.
Une durée plus courte augmente la pression financière.
6. Dependents

Pourquoi ?

Plus il y a de personnes à charge, plus les dépenses du ménage sont importantes.
7. Education

Pourquoi ?

Peut être corrélée au niveau de revenu et à la stabilité professionnelle.
8. Property_Area

Pourquoi ?

Certaines zones présentent des profils socio-économiques différents.
Variable cible

Loan_Status

Y = prêt accepté / faible risque
N = risque plus élevé

Accuracy=(TP+TN)/(TP+TN+FP+FN)

Précision

Rappel
Recall=TP/(TP+FN)
Mesure la capacité à détecter les mauvais payeurs


Score F1
F1=(2×Precision×Recall)/(Precision+Recall)

Exercice 5 : Stratégie d'évaluation pour différents modèles
1. Modèle supervisé : Random Forest
Métriques
Accuracy

Pourcentage global de bonnes prédictions.

Precision

Important lorsque les faux positifs coûtent cher.

Recall

Important pour détecter les défauts de paiement.

Score F1

Compromis entre précision et rappel.

AUC-ROC

Mesure la qualité globale du classifieur.

Méthodes d'évaluation
Validation croisée K-Fold
Diviser les données en K parties.
Entraîner K fois.
Calculer la moyenne des performances.

Matrice de confusion
|                | Réel Positif | Réel Négatif |
| -------------- | ------------ | ------------ |
| Prédit Positif | TP           | FP           |
| Prédit Négatif | FN           | TN           |


Courbe ROC

Visualise le compromis entre :

Taux de vrais positifs
Taux de faux positifs

Difficultés
Données déséquilibrées
Surapprentissage
Données manquantes

2. Modèle non supervisé : K-Means
Méthodes d'évaluation
Score de silhouette

Valeurs :

proche de 1 : excellent regroupement ;
proche de 0 : clusters peu distincts ;
négatif : mauvais regroupement.

Méthode du coude

Permet de déterminer le nombre optimal de clusters.

Davies-Bouldin Index

Mesure la séparation entre les groupes.

Difficultés
Absence de vérité terrain (labels).
Choix du nombre de clusters.
Sensibilité aux valeurs aberrantes.

3. Modèle d'apprentissage par renforcement : Q-Learning
Évaluation
Récompense cumulative

Somme totale des récompenses obtenues.

Taux de réussite

Pourcentage de fois où l'objectif est atteint.

Temps de convergence

Nombre d'épisodes nécessaires pour stabiliser la politique.

Exploration / Exploitation

Vérifier que l'agent :

explore suffisamment l'environnement ;
exploite ensuite les meilleures stratégies découvertes.
Difficultés
Convergence parfois lente.
Coût élevé des simulations.
Récompenses mal définies pouvant conduire à des comportements inattendus.

Conclusion

Pour la prédiction des défauts de paiement de prêts, les variables les plus importantes du jeu de données fourni sont principalement Credit_History, ApplicantIncome, CoapplicantIncome, LoanAmount et Loan_Amount_Term. Parmi les modèles, Random Forest et XGBoost constituent généralement les meilleurs choix grâce à leur capacité à capturer des relations complexes tout en offrant d'excellentes performances prédictives.

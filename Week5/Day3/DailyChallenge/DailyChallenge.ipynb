1. Charger l'ensemble de données MNIST
Importation des bibliothèques

```python
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Dense,
    Flatten,
    Conv2D,
    MaxPooling2D
)
```

```python
# Chargement des données
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
# Affichage des dimensions
print("X_train :", X_train.shape)
print("y_train :", y_train.shape)

print("X_test :", X_test.shape)
print("y_test :", y_test.shape)
```

```python
# Visualisation de quelques chiffres
plt.figure(figsize=(10,5))

for i in range(10):
    plt.subplot(2,5,i+1)
    plt.imshow(X_train[i], cmap="gray")
    plt.title(y_train[i])
    plt.axis("off")

plt.show()
```

2. Prétraitement pour le réseau neuronal entièrement connecté
Aplatir les images

```python
X_train_fc = X_train.reshape(60000, 784)
X_test_fc = X_test.reshape(10000, 784)
```

```python
# Normalisation
X_train_fc = X_train_fc.astype("float32") / 255
X_test_fc = X_test_fc.astype("float32") / 255
```

```python
# Encodage One-Hot
y_train_cat = to_categorical(y_train, 10)
y_test_cat = to_categorical(y_test, 10)
```

3. Construire et entraîner un réseau neuronal entièrement connecté
Création du modèle

```python
fc_model = Sequential()

fc_model.add(
    Dense(
        128,
        activation="relu",
        input_shape=(784,)
    )
)

fc_model.add(
    Dense(
        64,
        activation="relu"
    )
)

fc_model.add(
    Dense(
        10,
        activation="softmax"
    )
)
```

```python
# Résumé
fc_model.summary()
```

```python
# Compilation
fc_model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
# Entraînement

history_fc = fc_model.fit(
    X_train_fc,
    y_train_cat,
    epochs=10,
    batch_size=32,
    validation_split=0.2
)
```

```python
# Évaluation
loss_fc, accuracy_fc = fc_model.evaluate(
    X_test_fc,
    y_test_cat
)

print("Accuracy FC :", accuracy_fc)
```

```python
# 4. Prétraitement pour le CNN
# Reshape
X_train_cnn = X_train.reshape(
    60000,
    28,
    28,
    1
)

X_test_cnn = X_test.reshape(
    10000,
    28,
    28,
    1
)
```

```python
# Normalisation
X_train_cnn = X_train_cnn.astype("float32") / 255
X_test_cnn = X_test_cnn.astype("float32") / 255
```

```python
# One-Hot Encoding
y_train_cat = to_categorical(y_train,10)
y_test_cat = to_categorical(y_test,10)
```

5. Concevoir et entraîner un CNN
Création du modèle

```python
cnn_model = Sequential()
```

```python
# Couche convolutionnelle
cnn_model.add(
    Conv2D(
        filters=32,
        kernel_size=(3,3),
        activation="relu",
        input_shape=(28,28,1)
    )
)
```

```python
# Pooling
cnn_model.add(
    MaxPooling2D(
        pool_size=(2,2)
    )
)
```

```python
# Deuxième convolution
cnn_model.add(
    Conv2D(
        64,
        (3,3),
        activation="relu"
    )
)
```

```python
# Deuxième Pooling
cnn_model.add(
    MaxPooling2D(
        pool_size=(2,2)
    )
)
```

```python
# Flatten
cnn_model.add(
    Flatten()
)
```

```python
# Dense
cnn_model.add(
    Dense(
        128,
        activation="relu"
    )
)
```

```python
# Sortie
cnn_model.add(
    Dense(
        10,
        activation="softmax"
    )
)
```

```python
# Résumé
cnn_model.summary()
```

```python
# Compilation
cnn_model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
# Entraînement
history_cnn = cnn_model.fit(
    X_train_cnn,
    y_train_cat,
    epochs=10,
    batch_size=32,
    validation_split=0.2
)
```

```python
# Évaluation
loss_cnn, accuracy_cnn = cnn_model.evaluate(
    X_test_cnn,
    y_test_cat
)

print("Accuracy CNN :", accuracy_cnn)
```

```python
6. Comparaison des performances
Tableau comparatif
```

```python
import pandas as pd

results = pd.DataFrame({
    "Model":[
        "Fully Connected NN",
        "CNN"
    ],
    "Accuracy":[
        accuracy_fc,
        accuracy_cnn
    ]
})

print(results)
```

```python
# Graphique
import seaborn as sns

plt.figure(figsize=(6,4))

sns.barplot(
    data=results,
    x="Model",
    y="Accuracy"
)

plt.title("Comparaison des modèles")
plt.show()
```

Analyse des résultats
Réseau entièrement connecté
Traite chaque pixel indépendamment.
Ignore la structure spatiale de l'image.
Bonne performance (~97-98%).
Réseau convolutionnel (CNN)
Exploite les motifs locaux.
Détecte automatiquement :
contours,
formes,
courbes,
textures.
Généralement plus précis (~99%).
Conclusion
Cette étude a permis de comparer un réseau neuronal entièrement connecté à un réseau neuronal convolutionnel pour la reconnaissance des chiffres manuscrits du jeu de données MNIST.

Après prétraitement des données, les deux modèles ont été entraînés et évalués sur les mêmes données de test. Le réseau entièrement connecté a obtenu une précision élevée, généralement comprise entre 97 % et 98 %. Cependant, le réseau convolutionnel a obtenu de meilleures performances avec une précision proche de 99 %.

Cette différence s'explique par la capacité des couches convolutionnelles à extraire automatiquement les caractéristiques importantes des images tout en conservant leur structure spatiale. Les CNN sont donc particulièrement adaptés aux tâches de vision par ordinateur et constituent aujourd'hui la référence pour les problèmes de classification d'images.
